  - It finds the best hyperplane that separates the classes with the maximum margin.
2. What is the difference between Hard Margin and Soft Margin SVM?
- Hard Margin
  - No misclassification allowed.
  - Works only for linearly separable data.
  - Very sensitive to noise & outliers.
- Soft Margin
  - Allows some misclassification using a penalty term (C).
  - Works for non-perfectly separable data.
  - More robust to noise.

4. What is the role of Lagrange Multipliers in SVM?
  - They help convert the constrained optimization problem into a dual form that is easier to solve.
  - Only some multipliers are non-zero → these correspond to support vectors.
5. What are Support Vectors in SVM?
  - They are the critical data points closest to the hyperplane.
  - The decision boundary depends completely on them.
6. What is a Support Vector Classifier (SVC)?
  - SVC is the SVM model used for classification tasks.
7. What is a Support Vector Regressor (SVR)?
  - SVR is the SVM version used for regression tasks.
  - It uses an epsilon-insensitive loss function.
8. What is the Kernel Trick in SVM?
  - The kernel trick allows SVM to create non-linear decision boundaries by mapping inputs into a higher-dimensional space without explicitly computing that transformation.

9. Compare Linear Kernel, Polynomial Kernel, and RBF Kernel.
  - In Support Vector Machines, the choice of kernel determines how well the model can separate data. The Linear kernel is the simplest and is used when the data is linearly separable; it creates a straight-line (or flat) decision boundary and is very fast and efficient for high-dimensional datasets such as text classification. The Polynomial kernel, on the other hand, can capture more complex relationships by creating curved boundaries of different degrees; however, it can be computationally expensive and may overfit if the degree is too high. The RBF (Radial Basis Function) kernel is the most widely used because it can model highly non-linear patterns by mapping the data into an infinite-dimensional space, allowing very flexible decision boundaries. Overall, Linear kernels work best for simple relationships, Polynomial kernels for moderately complex patterns, and RBF kernels for highly non-linear datasets.

  10. What is the effect of the C parameter in SVM?
  - High C → low tolerance for errors → may overfit
  - Low C → high tolerance → smoother boundary, may underfit
  - C is a regularization parameter.
11. What is the role of the Gamma parameter in RBF Kernel SVM?
  - High gamma → small radius of influence → overfitting
  - Low gamma → large radius → underfitting
  - Gamma controls how far the influence of a single training example reaches.
12. What is the Naïve Bayes classifier, and why is it called "Naïve"?

It is called naïve because it assumes all features are independent, which is rarely true in real datasets.

14. Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes.
  - Naïve Bayes has different variants depending on the type of data being modeled. Gaussian Naïve Bayes is used when features are continuous and follow a normal (Gaussian) distribution, making it suitable for datasets like sensor readings or height-weight measurements. Multinomial Naïve Bayes is designed for count-based features, such as word frequencies in documents, which is why it is widely used in text classification problems like spam detection or topic modeling. Bernoulli Naïve Bayes, on the other hand, works with binary features, where values are either 0 or 1, indicating the absence or presence of a feature; this makes it useful for text classification with binary word indicators. Each variant uses Bayes’ theorem but applies it differently depending on the feature type.

19. Compare SVM and Naïve Bayes for classification tasks.
  - Support Vector Machines (SVM) and Naïve Bayes are both popular classification algorithms but differ significantly in their approach. SVM is a margin-based classifier that finds the best separating boundary between classes and performs well with both linear and non-linear data through the use of kernels. It generally offers high accuracy but can be computationally expensive and requires careful hyperparameter tuning. Naïve Bayes, in contrast, is a probabilistic classifier that assumes independence between features; this assumption makes the model extremely fast and efficient, especially for high-dimensional data such as text. While SVM typically performs better on complex datasets, Naïve Bayes often outperforms SVM in text classification due to its simplicity, speed, and ability to handle large vocabularies. Overall, SVM is better for complex decision boundaries, whereas Naïve Bayes is ideal for simpler or text-based classification tasks.

```python
#21. Python program to train an SVM Classifier on the iris dataset and evaluate accuracy.

import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load the dataset
iris = load_iris()
X, y = iris.data, iris.target

# 2. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Initialize and train the SVM Classifier
# We'll use a Linear Kernel for simplicity
svm_classifier = SVC(kernel='linear', random_state=42)
svm_classifier.fit(X_train, y_train)

# 4. Make predictions
y_pred = svm_classifier.predict(X_test)

# 5. Evaluate accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"--- SVM Classifier on Iris Dataset ---")
print(f"Accuracy: {accuracy:.4f}")
```

```python
#24. Python program to train an SVM Classifier with a Polynomial Kernel and visualize the decision boundary.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.svm import SVC

# 1. Generate a non-linear dataset
X, y = make_moons(n_samples=100, noise=0.15, random_state=42)

# 2. Train the SVM Classifier with a Polynomial Kernel
# C=10 and degree=3 are chosen for a clear non-linear boundary
poly_svm = SVC(kernel='poly', degree=3, C=10, random_state=42)
poly_svm.fit(X, y)

# 3. Visualization function
def plot_decision_boundary(model, X, y, title):
    # Set up the mesh grid
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                         np.arange(y_min, y_max, 0.02))

    # Predict the class for each point in the mesh
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # Plot the contour and the training points
    plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.RdBu)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.RdBu)
    plt.title(title)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.show()

# 4. Plot the result
print("--- Visualizing SVM with Polynomial Kernel ---")
plot_decision_boundary(poly_svm, X, y, "SVM Classifier with Polynomial Kernel (Degree 3)")
```

```python
#27. Write a Python program to train an SVM Classifier with different C values and compare the decision boundaries visually.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.svm import SVC
from matplotlib.colors import ListedColormap

# 1. Generate the dataset
X, y = make_moons(n_samples=100, noise=0.15, random_state=42)

# 2. Define C values to compare
C_values = [0.1, 1, 100]
titles = [f'C = {C}' for C in C_values]

# 3. Setup the plotting area
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
h = .02  # step size in the mesh

# Define the mesh grid
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_min, y_max, h))

# Custom colormaps
cm_bright = ListedColormap(['#FF0000', '#0000FF'])
cm = plt.cm.RdBu

# 4. Train and plot for each C value
for i, C in enumerate(C_values):
    # Initialize and train SVC with RBF kernel
    model = SVC(kernel='rbf', C=C, random_state=42)
    model.fit(X, y)

    # Predict on the mesh grid (Z)
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # Plot decision boundary (contour)
    axes[i].contourf(xx, yy, Z, cmap=cm, alpha=.8)

    # Plot the training points
    axes[i].scatter(X[:, 0], X[:, 1], c=y, cmap=cm_bright, edgecolors='k')

    axes[i].set_title(titles[i])
    axes[i].set_xticks(())
    axes[i].set_yticks(())

plt.suptitle("Effect of SVM Regularization Parameter C", fontsize=16)
plt.show()

# Interpretation:
# C=0.1 (Low): Large margin, allowing more misclassifications (smoother boundary).
# C=100 (High): Small margin, strict fit to training data (more complex/wiggly boundary, potential overfitting).
```

```python
#28. Write a Python program to train a Bernoulli Naive Bayes classifier for binary classification on a dataset with binary features.

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import BernoulliNB
from sklearn.metrics import accuracy_score
from sklearn.pipeline import make_pipeline

# 1. Load a subset of the dataset (binary classification)
categories = ['rec.autos', 'rec.sport.baseball']
newsgroups_train = fetch_20newsgroups(subset='train', categories=categories, random_state=42)
newsgroups_test = fetch_20newsgroups(subset='test', categories=categories, random_state=42)

# 2. Use CountVectorizer with binary=True for BernoulliNB
# binary=True ensures the feature vector contains only 0s (absent) or 1s (present)
vectorizer = CountVectorizer(binary=True)

# 3. Create and train the model
# BernoulliNB is ideal for binary features
model = make_pipeline(vectorizer, BernoulliNB())
model.fit(newsgroups_train.data, newsgroups_train.target)

# 4. Make predictions
y_pred = model.predict(newsgroups_test.data)

# 5. Evaluate accuracy
accuracy = accuracy_score(newsgroups_test.target, y_pred)

print(f"--- Bernoulli Naive Bayes (Binary Text Classification) ---")
print(f"Accuracy: {accuracy:.4f}")
print("Bernoulli NB only considers word presence (1) or absence (0).")
```

```python
#29. Write a Python program to apply feature scaling before training an SVM model and compare results with unscaled data.

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# 1. Load the dataset
wine = load_wine()
X, y = wine.data, wine.target

# 2. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- A. Unscaled Data ---
svc_unscaled = SVC(kernel='rbf', gamma='scale', random_state=42)
svc_unscaled.fit(X_train, y_train)
acc_unscaled = accuracy_score(y_test, svc_unscaled.predict(X_test))

# --- B. Scaled Data ---
# Initialize the Scaler
scaler = StandardScaler()

# Fit the scaler ONLY on the training data and transform both sets
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

svc_scaled = SVC(kernel='rbf', gamma='scale', random_state=42)
svc_scaled.fit(X_train_scaled, y_train)
acc_scaled = accuracy_score(y_test, svc_scaled.predict(X_test_scaled))

# 3. Compare Results
print(f"--- Feature Scaling Impact on SVM ---")
print(f"Accuracy (Unscaled Data): {acc_unscaled:.4f}")
print(f"Accuracy (Scaled Data):   {acc_scaled:.4f}")

# The scaled result should generally be significantly higher for the RBF kernel.
```

```python
#30. Write a Python program to train a Gaussian Naive Bayes model and compare the predictions before and after Laplace Smoothing

import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# 1. Load the dataset
iris = load_iris()
X, y = iris.data, iris.target

# 2. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- A. Gaussian NB with Default (Low) Smoothing ---
# var_smoothing is the portion of the largest variance of all features that is added to variances for calculation stability.
gnb_default = GaussianNB(var_smoothing=1e-9)
gnb_default.fit(X_train, y_train)
acc_default = accuracy_score(y_test, gnb_default.predict(X_test))

# --- B. Gaussian NB with Increased Smoothing (to show effect) ---
# A higher var_smoothing makes the model's estimate of the data distribution less dependent on the observed training variance.
gnb_smoothed = GaussianNB(var_smoothing=1e-5)
gnb_smoothed.fit(X_train, y_train)
acc_smoothed = accuracy_score(y_test, gnb_smoothed.predict(X_test))

# 3. Compare Results
print(f"--- Effect of Laplace/Variance Smoothing in Gaussian NB ---")
print(f"Accuracy (Default Smoothing): {acc_default:.4f} (var_smoothing=1e-9)")
print(f"Accuracy (Increased Smoothing): {acc_smoothed:.4f} (var_smoothing=1e-5)")

# Note: The difference might be small or zero on clean datasets like Iris,
# but smoothing is critical for stability and generalization on noisy/sparse data.
```

```python
#31. Write a Python program to train an SVM Classifier and use GridSearchCV to tune the hyperparameters (C, gamma, kernel).

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load the dataset
iris = load_iris()
X, y = iris.data, iris.target

# 2. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Define the parameter grid
# C: Regularization parameter. Small C -> wider margin, more errors. Large C -> narrower margin, fewer errors.
# gamma: Kernel coefficient for 'rbf', 'poly' and 'sigmoid'. Controls influence of single training example.
param_grid = {
    'C': [0.1, 1, 10],
    'gamma': [0.01, 0.1, 1],
    'kernel': ['rbf', 'linear']
}

# 4. Initialize GridSearchCV
# cv=5 means 5-fold cross-validation
grid_search = GridSearchCV(SVC(), param_grid, cv=5, verbose=2, scoring='accuracy')

# 5. Execute the search
print("--- Starting GridSearchCV for SVM Hyperparameter Tuning ---")
grid_search.fit(X_train, y_train)

# 6. Report the best parameters and score
print("\n--- Results ---")
print(f"Best parameters found: {grid_search.best_params_}")
print(f"Best cross-validation accuracy: {grid_search.best_score_:.4f}")

# 7. Evaluate on the test set
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)
test_accuracy = accuracy_score(y_test, y_pred)
print(f"Test Set Accuracy with Best Model: {test_accuracy:.4f}")
```

```python
#32. Write a Python program to train an SVM Classifier on an imbalanced dataset and apply class weighting and check if it improve accuracy.

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# 1. Create a highly imbalanced dataset (95% class 0, 5% class 1)
X, y = make_classification(n_samples=1000, n_features=10, n_redundant=0,
                           n_classes=2, n_clusters_per_class=1,
                           weights=[0.95, 0.05], flip_y=0, random_state=42)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- A. Unweighted SVM Model ---
svc_unweighted = SVC(kernel='linear', random_state=42)
svc_unweighted.fit(X_train, y_train)
y_pred_unweighted = svc_unweighted.predict(X_test)

# --- B. Class Weighted SVM Model ---
# 'balanced' automatically adjusts weights inversely proportional to class frequencies.
svc_weighted = SVC(kernel='linear', class_weight='balanced', random_state=42)
svc_weighted.fit(X_train, y_train)
y_pred_weighted = svc_weighted.predict(X_test)

# 3. Compare Results
print(f"--- SVM on Imbalanced Data (Class 1 is minority) ---")

print("\n[A] Unweighted Model Performance:")
print(classification_report(y_test, y_pred_unweighted, target_names=['Class 0 (Maj)', 'Class 1 (Min)']))

print("\n[B] Class Weighted Model Performance:")
print(classification_report(y_test, y_pred_weighted, target_names=['Class 0 (Maj)', 'Class 1 (Min)']))

# Note: The improvement will likely be seen in the minority class's Recall or F1-score.
```

```python
#35. Write a Python program to perform feature selection before training a Naive Bayes classifier and compare results.

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

# 1. Load the dataset (for text classification)
categories = ['comp.graphics', 'comp.os.ms-windows.misc', 'comp.sys.ibm.pc.hardware']
data_train = fetch_20newsgroups(subset='train', categories=categories, random_state=42)
data_test = fetch_20newsgroups(subset='test', categories=categories, random_state=42)
X_train, y_train = data_train.data, data_train.target
X_test, y_test = data_test.data, data_test.target

# --- A. Model without Feature Selection ---
pipe_full = Pipeline([
    ('vect', CountVectorizer()),
    ('mnb', MultinomialNB())
])
pipe_full.fit(X_train, y_train)
acc_full = accuracy_score(y_test, pipe_full.predict(X_test))


# --- B. Model with Feature Selection (SelectKBest) ---
# We select the top 1000 features based on chi-squared test
pipe_selected = Pipeline([
    ('vect', CountVectorizer()),
    ('feat_sel', SelectKBest(chi2, k=1000)), # Feature selection step
    ('mnb', MultinomialNB())
])
pipe_selected.fit(X_train, y_train)
acc_selected = accuracy_score(y_test, pipe_selected.predict(X_test))

# 3. Compare Results
print(f"--- Feature Selection Impact on Naive Bayes ---")
print(f"Accuracy (Full Feature Set):     {acc_full:.4f}")
print(f"Accuracy (Top 1000 Features):    {acc_selected:.4f}")
print("Feature selection can improve accuracy by reducing noise/overfitting.")
```

```python
#36. Write a Python program to train an SVM Classifier using One-vs-Rest (OvR) and One-vs-One (OvO) strategies on the Wine dataset and compare their accuracy.

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier
from sklearn.metrics import accuracy_score

# 1. Load the dataset (3 classes)
wine = load_wine()
X, y = wine.data, wine.target

# 2. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- A. One-vs-Rest (OvR) Strategy ---
# Using LinearSVC is common for OvR, but we'll use SVC for comparison consistency
ovr_classifier = OneVsRestClassifier(SVC(kernel='linear', random_state=42))
ovr_classifier.fit(X_train, y_train)
acc_ovr = accuracy_score(y_test, ovr_classifier.predict(X_test))

# --- B. One-vs-One (OvO) Strategy (Default for SVC) ---
# We wrap SVC explicitly for clarity, though it's the default behavior
ovo_classifier = OneVsOneClassifier(SVC(kernel='linear', random_state=42))
ovo_classifier.fit(X_train, y_train)
acc_ovo = accuracy_score(y_test, ovo_classifier.predict(X_test))

# 3. Compare Accuracies
print(f"--- Comparing SVM Multiclass Strategies ---")
print(f"Accuracy (One-vs-Rest - OvR): {acc_ovr:.4f}")
print(f"Accuracy (One-vs-One - OvO):  {acc_ovo:.4f}")
```

```python
#37. Write a Python program to train an SVM Classifier using Linear, Polynomial, and RBF kernels on the Breast Cancer dataset and compare their accuracy.

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# 1. Load and Scale the dataset (Scaling is crucial for RBF/Poly kernels)
cancer = load_breast_cancer()
X, y = cancer.data, cancer.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

kernels = ['linear', 'poly', 'rbf']
results = {}

print(f"--- Comparing SVM Kernel Performance ---")
for kernel in kernels:
    # Set degree=3 for Polynomial, keep others default
    degree = 3 if kernel == 'poly' else 1

    svc_model = SVC(kernel=kernel, degree=degree, random_state=42)
    svc_model.fit(X_train, y_train)

    y_pred = svc_model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    results[kernel] = accuracy

    print(f"Accuracy ({kernel.capitalize()} Kernel): {accuracy:.4f}")
```

```python
#38. Write a Python program to train an SVM Classifier using Stratified K-Fold Cross-Validation and compute the average accuracy.

from sklearn.datasets import load_wine
from sklearn.model_selection import StratifiedKFold
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np

# 1. Load the dataset
wine = load_wine()
X, y = wine.data, wine.target

# 2. Define the cross-validation strategy
# n_splits=5 is standard 5-fold CV
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
accuracies = []

# 3. Perform Cross-Validation
print(f"--- SVM with Stratified 5-Fold Cross-Validation ---")
for fold, (train_index, test_index) in enumerate(skf.split(X, y)):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]

    # Train the SVM model
    svc = SVC(kernel='linear', random_state=42)
    svc.fit(X_train, y_train)

    # Evaluate
    y_pred = svc.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)
    print(f"Fold {fold+1} Accuracy: {accuracy:.4f}")

# 4. Compute Average Accuracy
avg_accuracy = np.mean(accuracies)
print(f"\nAverage Cross-Validation Accuracy: {avg_accuracy:.4f}")
```

```python
#39. Write a Python program to train a Naive Bayes classifier using different prior probabilities and compare performance.

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# 1. Load the dataset
iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Calculate the actual prior probabilities from the training data
unique, counts = np.unique(y_train, return_counts=True)
actual_priors = counts / len(y_train)

# --- A. Default Priors (Learned from training data) ---
gnb_default = GaussianNB()
gnb_default.fit(X_train, y_train)
acc_default = accuracy_score(y_test, gnb_default.predict(X_test))

# --- B. Equal Priors (Assuming all classes are equally likely) ---
# Since Iris has 3 classes, equal prior is 1/3 for each.
equal_priors = np.array([1/3, 1/3, 1/3])
gnb_equal = GaussianNB(priors=equal_priors)
gnb_equal.fit(X_train, y_train)
acc_equal = accuracy_score(y_test, gnb_equal.predict(X_test))

# 3. Compare Results
print(f"--- Effect of Prior Probabilities in GNB ---")
print(f"Actual Priors: {actual_priors.round(2)}")
print(f"Accuracy (Actual/Default Priors): {acc_default:.4f}")
print(f"Accuracy (Equal Priors [1/3]):    {acc_equal:.4f}")
```

```python
#40. Write a Python program to perform Recursive Feature Elimination (RFE) before training an SVM classifier and compare accuracy.

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.feature_selection import RFE
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# 1. Load and Scale the dataset (30 features)
cancer = load_breast_cancer()
X, y = cancer.data, cancer.target
X_scaled = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# --- A. Model with Full Feature Set ---
svc_full = SVC(kernel="linear", random_state=42)
svc_full.fit(X_train, y_train)
acc_full = accuracy_score(y_test, svc_full.predict(X_test))

# --- B. Model with RFE-Selected Features ---
# Use LinearSVC as the estimator (often faster for RFE)
estimator = SVC(kernel="linear")
# Select the top 10 features
rfe = RFE(estimator, n_features_to_select=10, step=1)
rfe.fit(X_train, y_train)

# Transform the datasets to include only the selected features
X_train_rfe = rfe.transform(X_train)
X_test_rfe = rfe.transform(X_test)

# Train the final model on selected features
svc_rfe = SVC(kernel="linear", random_state=42)
svc_rfe.fit(X_train_rfe, y_train)
acc_rfe = accuracy_score(y_test, svc_rfe.predict(X_test_rfe))

# 3. Compare Results
print(f"--- Recursive Feature Elimination (RFE) with SVM ---")
print(f"Accuracy (Full Feature Set, 30 features): {acc_full:.4f}")
print(f"Accuracy (RFE-Selected, 10 features):   {acc_rfe:.4f}")
```

```python
#45. Write a Python program to train a Naive Bayes Classifier and evaluate its performance using the ROC-AUC score.

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

# Load dataset and split
cancer = load_breast_cancer()
X, y = cancer.data, cancer.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Gaussian Naive Bayes
gnb_model = GaussianNB()
gnb_model.fit(X_train, y_train)

# Get the probability of the positive class (class 1)
y_proba = gnb_model.predict_proba(X_test)[:, 1]

# Calculate ROC-AUC Score (Closer to 1.0 is better)
roc_auc = roc_auc_score(y_test, y_proba)

# Plot ROC Curve
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (1 - Specificity)')
plt.ylabel('True Positive Rate (Sensitivity/Recall)')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc="lower right")
plt.show()

print(f"--- Naive Bayes Evaluation with ROC-AUC ---")
print(f"ROC-AUC Score: {roc_auc:.4f}")
```

```python
#46. Write a python program to train an SVM classifier and visualise the precision - recall curve ?

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.metrics import precision_recall_curve, auc
from sklearn.preprocessing import StandardScaler

# 1. Load the dataset
# We use the Breast Cancer dataset, which is a binary classification task.
cancer = load_breast_cancer()
X, y = cancer.data, cancer.target

# 2. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Scale the features
# Scaling is essential for SVM models (both SVC and LinearSVC)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Initialize and Train the LinearSVC Model
# We use LinearSVC because it's faster and its decision_function is easily accessible
# for calculating PR curve scores.
svm_classifier = LinearSVC(random_state=42, dual='auto')
svm_classifier.fit(X_train_scaled, y_train)

# 5. Get the decision scores
# The decision_function returns the distance of each sample to the hyperplane.
# These scores are used as the probability-like confidence for the PR curve.
# For binary classification: score > 0 predicts class 1, score < 0 predicts class 0.
y_scores = svm_classifier.decision_function(X_test_scaled)

# 6. Compute Precision-Recall pairs for all thresholds
precision, recall, thresholds = precision_recall_curve(y_test, y_scores)

# 7. Calculate the Area Under the Curve (AUPRC / AUC)
# This gives a single metric summarizing the PR curve performance.
pr_auc = auc(recall, precision)

# 8. Visualize the Precision-Recall Curve
plt.figure(figsize=(8, 6))
plt.plot(recall, precision, color='darkorange', lw=2,
         label=f'LinearSVC (AUPRC = {pr_auc:.4f})')
plt.xlabel('Recall (True Positive Rate)')
plt.ylabel('Precision (Positive Predictive Value)')
plt.title('Precision-Recall Curve for SVM Classifier')
plt.legend(loc="lower left")
plt.grid(True)
plt.show()

print("--- SVM Classifier and Precision-Recall Curve ---")
print(f"Area Under the Precision-Recall Curve (AUPRC): {pr_auc:.4f}")
print("A value closer to 1.0 indicates better performance, especially on imbalanced datasets.")
```
